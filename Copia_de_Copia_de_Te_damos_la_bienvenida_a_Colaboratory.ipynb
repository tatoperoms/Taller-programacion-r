```r
# =========================
# PUNTO 1
# =========================

# ---------
# Punto 1.a
# ---------

x <- 5

primero  <- (3*x^7 - 5*x^6 + 2*x^4 - 9*x + 12)/(4*x^2 - 7)
segundo  <- 6*x^5 - (2*x^3 - 5*x - 1)^2
tercero  <- 4*x^2*(x - 3)*(x + 5)
cuarto   <- ((x^2 - 4)^3)/(2*x - 1) + 7*x^4
quinto   <- -((3*x - 2)/(x + 1))^2 + (x^3 - 2*x + 1)^3
sexto    <- -5*(x - 2)^4 + (9*x^5 - 3*x^3 + x - 8)/(x^2 + 2)
septimo  <- -2*x^2*(x^2 - 1)*(x + 4) + 10

resultado1a <- primero + segundo + tercero + cuarto + quinto + sexto + septimo
cat("Resultado Punto 1.a:\n")
print(resultado1a)


# ---------
# Punto 1.b
# ---------

p1  <- 8*x^6
p2  <- -(5*x^8 - 3*x^5 + 2*x^2 - 7)/(x^2 - 3)
p3  <- (3*x^2 - 4*x + 1)^3
p4  <- -6*x*(x - 2)^4
p5  <- (2*x^3 - x + 5)^2/(4*x - 1)
p6  <- -9*x^5
p7  <- 7*(x^2 + 1)*(x - 3)*(x + 2)
p8  <- -((x^2 - 1)/(x + 4))^3
p9  <- 12*x^4 - 2*(x - 1)^5
p10 <- (6*x^7 - x^4 + 9*x - 10)/(x^2 + 5) - 15

resultado1b <- p1 + p2 + p3 + p4 + p5 + p6 + p7 + p8 + p9 + p10
cat("\nResultado Punto 1.b:\n")
print(resultado1b)


# =========================
# PUNTO 2
# =========================

a <- TRUE
b <- FALSE

resultado2a <-
((!a & b) | (a & !b) | (!(a & b) & (a | b))) &
((a | !b) & (!(!a | b)) | (!b & a & a)) |
(!((a & b) | (!a & !b) | (b & !a)) & (a | !a)) |
((!(b | b) & (a & a)) & !(!a & !b))

cat("\nResultado Punto 2.a:\n")
print(resultado2a)


resultado2b <-
(a & a & !b) | (!a & b) |
!(a & !b & b) &  ((a | b) | !(!a | !b)) |
(!((a & b) | (!a & b) | (!b & !b)) & (a | !(b & a))) |
(!(a | !a) | (a & !(b | b))) & (!(!(!a)) | (!b & a))

cat("\nResultado Punto 2.b:\n")
print(resultado2b)


# =========================
# PUNTO 3
# =========================

A  <- "Árboles"
B  <- "Bosques"
C  <- "Casas"
D  <- "Patios"
E  <- "Perros"
F  <- "Gatos"
G  <- "estará"
H  <- "Quiere"
I  <- "Comer"
J  <- "Mucho"
K  <- "Solo"
L  <- "Dormir"
N  <- "Correr"
O  <- "Jugar"
P  <- "Carne"
Q  <- "Pollo"

R  <- "Los"
S  <- "y"
T  <- "rodean"
U  <- "las"
V  <- "del"
W  <- "lugar"
X  <- "En"
Y  <- "esa"
Z  <- "zona"
AA <- "viven"
AB <- "muy"
AC <- "lindos"
AD <- "Cada"
AE <- "uno"
AF <- "la"
AG <- "el"
AH <- "son"
AI <- "sus"
AJ <- "favoritos"
AK <- "El"
AL <- "silencio"
AM <- "ensordecedor"
AN <- "a"
AO <- "veces"
AP <- "los"
AQ <- "deja"
AR <- "y"
AS <- "durante"
AT <- "la"
AU <- "tarde"
AV <- "tranquilo"
AW <- "para"
AX <- "disfrutar"
AY <- "de"
AZ <- "cerca"

sub_dorm <- substr(L, 1, 6)
sub_corr <- substr(N, 1, 6)

texto <- paste0(
R, " ", tolower(A), " ", S, " ", tolower(B), " ", T, " ", U, " ", tolower(C), " ", S, " ", tolower(D), " ", V, " ", W, ".\n",
X, " ", Y, " ", Z, " ", AA, " ", tolower(E), " ", S, " ", tolower(F), " ", AB, " ", AC, ".\n",
AD, " ", AE, " ", tolower(H), " ", tolower(I), " ", tolower(J), ", ", AF, " ", tolower(P), " ", S, " ", AG, " ", tolower(Q), " ", AH, " ", AI, " ", AJ, ".\n",
AK, " ", AL, " ", AM, " ", AN, " ", AO, " ", AP, " ", AQ, " ", tolower(sub_dorm), ", ",
AR, " ", AS, " ", AT, " ", AU, " ", tolower(G), " ", AV, " ", AW, " ", AX, " ", AY, " ",
tolower(K), " ", tolower(sub_corr), " ", S, " ", tolower(O), " ", AZ, "."
)

cat("\nTexto completo:\n\n")
cat(texto)

cat("\n\nCantidad de caracteres:\n")
print(nchar(texto))


# =========================
# PUNTO 4
# =========================

cat("\nTipo de dato Punto 1.a:\n")
print(class(resultado1a))

cat("\nTipo de dato Punto 1.b:\n")
print(class(resultado1b))

cat("\nTipo de dato Punto 2.a:\n")
print(class(resultado2a))

cat("\nTipo de dato Punto 2.b:\n")
print(class(resultado2b))

cat("\nTipo de dato del texto:\n")
print(class(texto))
```
